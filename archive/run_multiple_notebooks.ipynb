```python
import papermill as pm
import json
import os
import pandas as pd

# Anzahl der Durchläufe
num_runs = 30

# Liste, in der die Ergebnisse aller Durchläufe gespeichert werden
all_results = []

# Stelle sicher, dass beim Start keine alte results.json vorhanden ist.
if os.path.exists("results.json"):
    os.remove("results.json")

# Schleife: Führe das Notebook mehrfach aus.
for i in range(num_runs):
    print(f"Run {i+1}/{num_runs} ...")
    
    # Erstelle einen eindeutigen Namen für das ausgeführte Notebook (Output-Notebook)
    output_notebook = f"output_run_{i+1}.ipynb"
    
    
    # Notebook ausführen
    pm.execute_notebook(
        "load_profile_prediciton_2.ipynb",  # Pfad zum Notebook
        output_notebook,  # Ausgabedatei

    )
    
    # Nachdem das Notebook ausgeführt wurde, wird angenommen, dass es "results.json" geschrieben hat.
    if os.path.exists("results.json"):
        with open("results.json", "r") as f:
            result = json.load(f)
            # Convert any Timestamp objects to strings
            for key, value in result.items():
                if isinstance(value, pd.Timestamp):
                    result[key] = value.isoformat()
                elif isinstance(value, list):
                    result[key] = [v.isoformat() if isinstance(v, pd.Timestamp) else v for v in value]
            all_results.append(result)
        # Optional: Lösche die Datei, damit sie im nächsten Durchlauf frisch geschrieben wird.
        os.remove("results.json")
    else:
        print("WARNUNG: Es wurde keine results.json gefunden!")
    
    print(f"Ergebnis für Lauf {i+1}: {result}\n")

# Speichere alle Ergebnisse in einer Datei (optional)
with open("all_runs_results.json", "w") as f:
    json.dump(all_results, f, indent=4)

print("Alle Ergebnisse:")
print(all_results)
```

```python
#Compare results
with open("all_runs_results_14days.json", "r") as f:
    all_results = json.load(f)

# 2. Übersicht: Ausgabe des Run-Datums und der durchschnittlichen NMAE pro Modell
print("Übersicht der Runs (Run Date und Average NMAE):")
for run in all_results:
    run_date = run["Run Date"]
    avg_nmae = run["Average NMAE"]
    print(f"Run Date: {run_date}")
    for model, value in avg_nmae.items():
        print(f"  {model}: {value:.2f}")
    print()
```

```python
import matplotlib.pyplot as plt
from datetime import datetime
import numpy as np
```

1. JSON‑Datei mit den Ergebnissen laden und nmae

```python
run_dates = [datetime.fromisoformat(run["Run Date"]) for run in all_results]
avg_nmae_model1 = [run["Average NMAE"]["Model 1"] for run in all_results]

avg_nmae_model2 = [run["Average NMAE"]["Model 2"] for run in all_results]

avg_nmae_model3 = [run["Average NMAE"]["Model 3"] for run in all_results]

plt.figure(figsize=(10, 6))
plt.scatter(run_dates, avg_nmae_model1, label="Model 1")
plt.scatter(run_dates, avg_nmae_model2, label="Model 2")
plt.scatter(run_dates, avg_nmae_model3, label="Model 3")
plt.xlabel("Run Date")
plt.ylabel("Average NMAE (%)")
plt.title("Durchschnittliche NMAE pro Modell über die Runs - 4 Wochen Haushaltsbeobachtung in 2019")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

4. Vergleich: Durchschnittliche jährliche Kosten (Predicted vs. Actual)

```python
print("Vergleich: Durchschnittliche jährliche Kosten (Predicted vs. Actual)")
for run in all_results:
    run_date = run["Run Date"]
    print(f"Run Date: {run_date}")
    
    # Actual costs sind in einer Liste
    actual_costs = run["Annual Predicted Costs"]["Actual Costs"]
    avg_actual = np.mean(actual_costs)
    
    for model in ["Model 1", "Model 2", "Model 3"]:
        predicted_costs = run["Annual Predicted Costs"][model]
        avg_predicted = np.mean(predicted_costs)
        diff = avg_predicted - avg_actual
        print(f"  {model}: Predicted = {avg_predicted:.2f} €, Actual = {avg_actual:.2f} €, Diff = {diff:+.2f} €")
    print()

# 4. Visualisierung: Vergleich der durchschnittlichen jährlichen Kosten (Predicted vs. Actual)
plt.figure(figsize=(10, 6))
bar_width = 0.25
index = np.arange(len(all_results))
plt.axhline(y=874.63, color='red', linestyle='--', label="Actual Costs")
bar2 = plt.bar(index + bar_width, [np.mean(run["Annual Predicted Costs"]["Model 1"]) for run in all_results], bar_width, label="Model 1")
bar3 = plt.bar(index + 2*bar_width, [np.mean(run["Annual Predicted Costs"]["Model 2"]) for run in all_results], bar_width, label="Model 2")
bar4 = plt.bar(index + 3*bar_width, [np.mean(run["Annual Predicted Costs"]["Model 3"]) for run in all_results], bar_width, label="Model 3")
plt.xlabel("Run Date")
plt.ylabel("Average Annual Costs (€)")
plt.yticks(np.arange(870, 890, 0.2))
plt.ylim(873, 878)
plt.title("Comparison of Average Annual Costs (Predicted vs. Actual)")
plt.xticks(index + 1.5*bar_width, run_dates, rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
